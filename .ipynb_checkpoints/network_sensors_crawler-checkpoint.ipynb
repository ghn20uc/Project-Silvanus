# GeoNet Network Sensors Crawler

Crawl GeoNet's `/network/sensors` endpoint for every known (station, sensorType) pair
and save the results to CSV.

**Pipeline**
1. `get_sensor_type_lookup()` — query `/network/sensor/type` once, build a `{Description: SensorType}` map (e.g. `{"Air pressure sensor": "1", "GNSS/GPS": "8"}`). Needed because `/network/station` returns `SensorType` as a **text label**, while `/network/sensors` requires the **numeric code**.
2. `get_station_sensor_ids()` — query `/network/station` once, get every `(station code, sensorType CODE)` pair that has ever existed (no `startDate`/`endDate` filter, so decommissioned stations are included too), save as a `.dat` ID file.
3. `flatten_sensor_response()` — the schema differs per sensorType (e.g. GNSS stations return `antenna` and `receiver` arrays with equipment-change history; other types may use different array/field shapes) - auto-detects scalar metadata vs. equipment records instead of hardcoding field names.
4. `fetch_sensor()` — async GET `/network/sensors` for one pair.
5. `gather_sensors()` — run `fetch_sensor()` concurrently for every pair, with a semaphore + bounded connector to avoid hammering the API.
6. `save_sensors_to_csv()` — flatten everything into one CSV (columns built dynamically).

Install deps first:
```
pip install httpx pandas requests --break-system-packages
```

**Note:** this uses `httpx` (not `aiohttp`) for the async requests. `aiohttp`'s internal
timeout handling has known conflicts with `nest_asyncio` / newer Python versions
(shows up as `Timeout context manager should be used inside a task`), so `httpx`
is the safer choice here — and modern Jupyter supports top-level `await` natively,
so `nest_asyncio` isn't needed either.


````python
import os
import csv
import asyncio
import httpx
import requests
import pandas as pd

API_URL = "https://api.geonet.org.nz"
IDFOLDER = "./IDS/"
OUTPUT_FOLDER = "./network_sensor_data/"

# Same global error-tracking pattern as functions.py / main.ipynb
error_count = 0
error_list = []
````

## Step 1-2: build the (station, sensorType CODE) ID list

````python
def get_sensor_type_lookup():
    """
    Query /network/sensor/type once and return {Description: SensorType} map,
    e.g. {"Air pressure sensor": "1", "GNSS/GPS": "8", ...}.
    Needed because /network/station returns SensorType as a text label,
    while /network/sensors requires the numeric code.
    """
    resp = requests.get(f"{API_URL}/network/sensor/type", timeout=30)
    resp.raise_for_status()
    data = resp.json()
    return {item["Description"]: item["SensorType"] for item in data.get("SensorTypes", [])}


def get_station_sensor_ids(idfolder, sensor_types=range(1, 14)):
    """
    Query GeoNet /network/station ONCE to get every (station code, sensorType
    CODE) pair, including decommissioned stations, and save them into a
    .dat ID file.

    Input
    idfolder : str
        Folder to save the generated ID file.
    sensor_types : iterable of int, optional
        Sensor type codes to include (default: 1-13, i.e. all known types).

    Output
    pairs_df : pandas.DataFrame
        DataFrame with columns ['Code', 'SensorType'] (SensorType is the
        numeric code) - the same pairs written to the .dat file.
    """
    os.makedirs(idfolder, exist_ok=True)
    types_str = ",".join(str(t) for t in sensor_types)
    name_to_code = get_sensor_type_lookup()

    url = f"{API_URL}/network/station"
    headers = {"Accept": "application/vnd.geo+json;version=2"}
    # No startDate/endDate -> include historical / decommissioned stations too
    params = {"sensorType": types_str}

    resp = requests.get(url, params=params, headers=headers, timeout=30)
    resp.raise_for_status()
    data = resp.json()

    rows = []
    skipped = 0
    for f in data.get("features", []):
        props = f.get("properties", {})
        code = props.get("Code")
        stype_name = props.get("SensorType")
        stype_code = name_to_code.get(stype_name)
        if code and stype_code:
            rows.append({"Code": code, "SensorType": stype_code})
        else:
            skipped += 1

    if skipped:
        print(f"⚠️ Skipped {skipped} station(s) with unrecognised SensorType label.")

    pairs_df = pd.DataFrame(rows).drop_duplicates().reset_index(drop=True)

    fnameout = os.path.join(idfolder, "NZ_network_sensor_ID.dat")
    pairs_df.to_csv(fnameout, index=False, header=None)

    return pairs_df


def load_station_sensor_ids(idfolder):
    """
    Load a previously-saved (station, sensorType) ID file instead of
    re-querying /network/station.

    Input
    idfolder : str
        Folder containing NZ_network_sensor_ID.dat

    Output
    pairs_df : pandas.DataFrame
        DataFrame with columns ['Code', 'SensorType'].
    """
    fname = os.path.join(idfolder, "NZ_network_sensor_ID.dat")
    return pd.read_csv(fname, header=None, names=["Code", "SensorType"])
````

## Step 3-4: fetch + flatten one (station, sensorType) pair

````python
def flatten_sensor_response(data, station, sensor_type):
    """
    Flatten one /network/sensors response into a list of flat rows.

    The schema differs per sensorType - e.g. GNSS stations return
    'antenna' and 'receiver' arrays (equipment-change history), while
    other sensor types may use different array names. Rather than
    hardcoding field names, this function auto-detects:
      - scalar fields (code, network, name, latitude, longitude,
        elevation, datum, ...)                      -> station-level metadata
      - list-of-dict fields (antenna, receiver, ...) -> equipment history
      - single dict fields (not wrapped in a list)   -> treated as one
        equipment record too, instead of being silently dropped

    and produces one row per equipment-history entry, tagged with which
    field it came from via 'equipment_category'. If an equipment field
    name collides with a station-level metadata field (e.g. both the
    station and its antenna have 'start-date'/'end-date'), the equipment
    field is renamed to '{group_name}_{field}' (e.g. 'antenna_start-date')
    instead of silently overwriting the station's value. If there are no
    equipment sub-fields at all, it returns a single row of just the metadata.

    Input
    data : dict
        Raw JSON response for one (station, sensorType) pair.
    station : str
    sensor_type : str or int

    Output
    list of dict
    """
    if not isinstance(data, dict):
        return []

    meta = {}
    equipment_groups = {}
    for k, v in data.items():
        if isinstance(v, list) and v and isinstance(v[0], dict):
            equipment_groups[k] = v
        elif isinstance(v, dict):
            # single equipment record (not wrapped in a list) - keep it too
            equipment_groups[k] = [v]
        elif isinstance(v, list):
            # list of scalars (e.g. list of strings) - flatten into one string
            meta[k] = ";".join(str(x) for x in v)
        else:
            meta[k] = v

    meta["QueriedCode"] = station
    meta["QueriedSensorType"] = sensor_type

    rows = []
    if equipment_groups:
        for group_name, entries in equipment_groups.items():
            for entry in entries:
                row = dict(meta)
                row["equipment_category"] = group_name
                for k, v in entry.items():
                    key = f"{group_name}_{k}" if k in row else k
                    row[key] = v
                rows.append(row)
    else:
        rows.append(meta)

    return rows


async def fetch_sensor(client, station, sensor_type, semaphore):
    """
    Fetch sensor detail for a single (station, sensorType) pair.

    Input
    client : httpx.AsyncClient
    station : str
    sensor_type : str or int
    semaphore : asyncio.Semaphore
        Limits concurrent requests so we don't hammer the API.

    Output
    list of dict, or None
        Flattened record(s) via flatten_sensor_response(), or None if the
        request failed or returned no data.
    """
    global error_count

    url = f"{API_URL}/network/sensors"
    params = {"sensorType": sensor_type, "station": station}
    headers = {"Accept": "application/json;version=2"}

    async with semaphore:
        try:
            response = await client.get(url, params=params, headers=headers)

            if response.status_code in (400, 404):
                # 404: khong ton tai. 400: thuong gap voi sensorType=10
                # (Manual collection) - loai nay khong co ho so thiet bi
                # nen GeoNet tra ve 400 thay vi 404. Coi ca hai la
                # "khong co du lieu", khong tinh la loi that.
                return None

            if response.status_code != 200:
                error_count += 1
                error_list.append(
                    f"{error_count}. Error fetching {station}/{sensor_type}: HTTP {response.status_code}"
                )
                return None

            data = response.json()
            out = flatten_sensor_response(data, station, sensor_type)
            return out if out else None

        except Exception as e:
            error_count += 1
            error_list.append(f"{error_count}. Error fetching {station}/{sensor_type}: {e}")
            return None
````

## Step 5: run `fetch_sensor()` concurrently for every pair

````python
async def gather_sensors(pairs_df, max_concurrent=8):
    """
    Fetch sensor details for all (station, sensorType) pairs concurrently.

    Input
    pairs_df : pandas.DataFrame
        DataFrame with columns ['Code', 'SensorType'].
    max_concurrent : int, optional
        Max number of concurrent requests (default: 8). GeoNet doesn't
        publish an official rate limit, keep this conservative.

    Output
    list of dict
        Flattened list of sensor detail records (failed/empty entries removed).
    """
    semaphore = asyncio.Semaphore(max_concurrent)
    limits = httpx.Limits(max_connections=max_concurrent, max_keepalive_connections=max_concurrent)
    timeout = httpx.Timeout(30.0)

    async with httpx.AsyncClient(limits=limits, timeout=timeout) as client:
        tasks = [
            asyncio.create_task(fetch_sensor(client, row.Code, row.SensorType, semaphore))
            for row in pairs_df.itertuples()
        ]
        results = await asyncio.gather(*tasks)

    flattened = []
    for r in results:
        if r:
            flattened.extend(r)
    return flattened
````

## Step 6: save everything to CSV

````python
def save_sensors_to_csv(data_list, filename, output_folder):
    """
    Save a list of sensor detail dicts to a CSV file.

    Input
    data_list : list of dict
        Output of gather_sensors().
    filename : str
    output_folder : str

    Notes
    -----
    Columns are built dynamically from whatever keys appear across all
    records, since different sensor types produce different equipment fields.
    """
    os.makedirs(output_folder, exist_ok=True)
    filepath = os.path.join(output_folder, filename)

    if not data_list:
        print(f"⚠️ No data to save for {filename}")
        return

    fieldnames = []
    for row in data_list:
        for k in row.keys():
            if k not in fieldnames:
                fieldnames.append(k)

    with open(filepath, "w", newline="", encoding="utf-8") as csvfile:
        writer = csv.DictWriter(csvfile, fieldnames=fieldnames)
        writer.writeheader()
        for row in data_list:
            writer.writerow(row)
````

## Run the pipeline

````python
os.makedirs(IDFOLDER, exist_ok=True)
os.makedirs(OUTPUT_FOLDER, exist_ok=True)

# Step 1-2: get every (station, sensorType code) pair, save as .dat ID file
pairs_df = get_station_sensor_ids(IDFOLDER)
# pairs_df = load_station_sensor_ids(IDFOLDER)  # <- use this instead to skip re-fetching
print(f"📂 Found {len(pairs_df)} (station, sensorType) pairs.")
````

````python
# Step 3-5: crawl sensor detail for every pair concurrently
print("⚙️ Fetching sensor details...")
data_list = await gather_sensors(pairs_df)
print(f"Got {len(data_list)} flattened records.")
````

````python
# Step 6: save results
save_sensors_to_csv(
    data_list,
    filename="network_sensors.csv",
    output_folder=OUTPUT_FOLDER,
)
print(f"✅ Saved {len(data_list)} sensor records to {OUTPUT_FOLDER}")

if error_list:
    print("\n⚠️ Some errors occurred:")
    for err in error_list:
        print(err)
else:
    print("\nAll done with no errors!")
````
